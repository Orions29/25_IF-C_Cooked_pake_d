# Cross-Model Comparison dan Trade-off Analysis

## 1. Gambaran Umum

Proyek ini mengevaluasi beberapa pendekatan machine learning untuk memprediksi permintaan pickup taksi NYC Yellow Taxi. Model yang digunakan terdiri dari tiga pendekatan utama, yaitu regresi, time-series forecasting, dan klasifikasi. Setiap pendekatan memiliki tujuan prediksi yang berbeda sehingga memiliki kelebihan, keterbatasan, dan metrik evaluasi masing-masing.

Model regresi digunakan untuk mengestimasi jumlah pickup secara numerik, model time-series digunakan untuk menangkap pola temporal dan musiman, sedangkan model klasifikasi digunakan untuk mengategorikan tingkat permintaan menjadi Rendah, Sedang, dan Tinggi. Oleh karena itu, perbandingan model dilakukan dengan mempertimbangkan performa prediksi, efisiensi komputasi, kompleksitas, dan kesesuaian terhadap tujuan penggunaannya, bukan dengan membandingkan metrik yang berbeda secara langsung.

---

## 2. Perbandingan Performa Model

### 2.1 Model Regresi

Member 2 mengevaluasi Linear Regression, Ridge Regression, dan Lasso Regression menggunakan data validation. Model awal menghasilkan performa yang relatif serupa, yang menunjukkan bahwa pendekatan linear mampu menangkap sebagian besar hubungan antara fitur yang digunakan dengan jumlah pickup.

| Model | MAE | RMSE | WAPE | R² | Training Time (s) |
|---|---:|---:|---:|---:|---:|
| Linear Regression | 4.1636 | 11.6050 | 20.37% | 0.9584 | 2.72 |
| Ridge Regression | 4.1636 | 11.6049 | 20.37% | 0.9584 | 1.39 |
| Lasso Regression | **4.1556** | 11.6513 | **20.33%** | 0.9580 | 3.72 |

Model Lasso awal menghasilkan MAE dan WAPE terendah di antara model regresi awal, sedangkan Ridge dan Linear Regression menghasilkan RMSE yang sedikit lebih rendah. Perbedaan performa di antara ketiga model relatif kecil.

Setelah hyperparameter tuning, Ridge Regression dengan alpha = 1000 menghasilkan MAE sebesar 4.1608 dan RMSE sebesar 11.6007. Sementara itu, Lasso dengan alpha = 1.0 menghasilkan MAE sebesar 4.2395 dan RMSE sebesar 11.6552.

Secara keseluruhan, model regresi menunjukkan performa yang relatif konsisten. Ridge memberikan keseimbangan yang kompetitif antara akurasi prediksi dan efisiensi komputasi.

### 2.2 Model Time-Series Forecasting

Member 3 mengevaluasi Seasonal Naive, Holt-Winters, dan SARIMA pada data pickup demand tingkat city-level.

| Model | MAE | RMSE | WAPE | R² | Fit Time (s) |
|---|---:|---:|---:|---:|---:|
| Seasonal Naive (168) | 806.4288 | 1,153.7025 | 15.00% | 0.8359 | 0.00 |
| Holt-Winters (A,N,A) | 952.8597 | 1,378.4181 | 17.73% | 0.7658 | 0.72 |
| SARIMA (1,0,1)(1,1,1,24) | **410.7729** | **585.4257** | **7.64%** | **0.9578** | 19.66 |

SARIMA menghasilkan performa validation terbaik di antara model time-series yang dievaluasi. Model ini menghasilkan MAE, RMSE, dan WAPE terendah serta R² tertinggi.

Meskipun waktu fitting SARIMA lebih lama dibandingkan Holt-Winters dan Seasonal Naive, peningkatan akurasi prediksi yang dihasilkan cukup signifikan. Hal tersebut menunjukkan bahwa tambahan biaya komputasi dapat dipertimbangkan ketika akurasi forecasting menjadi prioritas utama.

### 2.3 Model Klasifikasi

Member 4 mengevaluasi Linear SVM dan RBF SVM, termasuk konfigurasi baseline dan optimized. Macro F1 digunakan sebagai metrik evaluasi utama karena mempertimbangkan performa klasifikasi pada kelas demand Rendah, Sedang, dan Tinggi.

| Model | Accuracy | Macro Precision | Macro Recall | Macro F1 | Training Time (s) |
|---|---:|---:|---:|---:|
| Linear SVM Baseline | 73.96% | 73.44% | 72.96% | 3.13 |
| Linear SVM Optimized (C=10) | 73.89% | 73.24% | 72.80% | 3.19 |
| RBF SVM Baseline | 73.12% | 74.94% | 73.11% | 7.25 |
| RBF SVM Optimized (C=10, gamma=0.01) | **75.57%** | **76.67%** | **74.98%** | **6.66** |

Optimized RBF SVM menghasilkan Macro F1 dan accuracy tertinggi di antara konfigurasi klasifikasi yang dievaluasi. Hal ini menunjukkan bahwa kernel nonlinear RBF mampu menangkap hubungan antarfitur dan kategori demand lebih baik dibandingkan model linear dalam eksperimen ini.

Namun, RBF SVM membutuhkan waktu prediksi yang jauh lebih tinggi dibandingkan Linear SVM. Oleh karena itu, efisiensi komputasi tetap menjadi pertimbangan dalam pemilihan model klasifikasi.

## 3. Analisis Trade-off Antar Model

### 3.1 Perbedaan Tujuan Prediksi

Setiap pendekatan memiliki tujuan prediksi dan representasi target yang berbeda.

| Pendekatan | Model | Output Prediksi | Tujuan |
|---|---|---|---|
| Regresi | Linear, Ridge, Lasso | Jumlah pickup numerik | Mengestimasi jumlah demand |
| Time-Series Forecasting | Seasonal Naive, Holt-Winters, SARIMA | Demand numerik pada waktu mendatang | Menangkap pola temporal dan musiman |
| Klasifikasi | Linear SVM, RBF SVM | Rendah, Sedang, Tinggi | Mengategorikan tingkat demand |

Model regresi dan forecasting memberikan estimasi numerik, sedangkan model klasifikasi memberikan kategori demand. Pemilihan model bergantung pada kebutuhan sistem, apakah membutuhkan estimasi jumlah pickup, prediksi demand pada periode mendatang, atau pengelompokan tingkat demand.

### 3.2 Akurasi vs Kompleksitas Komputasi

Hasil evaluasi menunjukkan adanya trade-off antara performa prediksi dan kebutuhan komputasi.

Model regresi menghasilkan error prediksi numerik yang relatif rendah dengan waktu training yang cukup singkat. SARIMA menunjukkan performa forecasting city-level yang jauh lebih baik dibandingkan model time-series lainnya, tetapi membutuhkan waktu fitting lebih lama.

Pada klasifikasi, optimized RBF SVM menghasilkan Macro F1 tertinggi, tetapi membutuhkan waktu prediksi yang lebih besar dibandingkan Linear SVM.

Hal tersebut menunjukkan bahwa model yang lebih kompleks dapat memberikan peningkatan performa, tetapi manfaatnya perlu dibandingkan dengan biaya komputasi dan kebutuhan implementasi.

### 3.3 Interpretabilitas vs Performa

Linear Regression, Ridge, Lasso, dan Linear SVM umumnya memiliki struktur model yang lebih sederhana dibandingkan RBF SVM. Struktur yang lebih sederhana dapat mempermudah analisis hubungan antara fitur input dan hasil prediksi.

SARIMA memiliki komponen temporal yang dapat dianalisis melalui struktur autoregressive, differencing, moving average, dan seasonal. Namun, pemilihan konfigurasi serta interpretasi parameter memerlukan pemahaman time-series.

RBF SVM memberikan fleksibilitas lebih tinggi dalam menangkap hubungan nonlinear, tetapi proses pengambilan keputusannya lebih sulit dijelaskan secara langsung dibandingkan model linear.

Oleh karena itu, pemilihan model perlu mempertimbangkan keseimbangan antara performa prediksi dan kemudahan interpretasi.

### 3.4 Precision vs Recall

Hasil klasifikasi menunjukkan pentingnya mengevaluasi model menggunakan beberapa metrik, bukan hanya accuracy. Macro Precision, Macro Recall, dan Macro F1 memberikan gambaran performa model pada seluruh kategori demand.

Macro F1 digunakan sebagai metrik utama klasifikasi karena mempertimbangkan precision dan recall dari setiap kelas. Hal ini penting agar evaluasi tidak hanya didominasi oleh kelas yang memiliki jumlah data lebih banyak.

## 4. Analisis Coverage@K (RQ4)

Coverage@K digunakan untuk mengevaluasi seberapa besar pickup aktual yang dapat tercakup ketika hanya K zona dengan prioritas tertinggi dipilih setiap jam. Analisis ini mensimulasikan kondisi ketika sumber daya terbatas dan hanya sejumlah zona yang dapat diprioritaskan.

Strategi prioritas berbasis SVM dibandingkan dengan baseline lag-1, yang menentukan prioritas zona berdasarkan pickup pada periode sebelumnya.

| K | Coverage SVM | Coverage Baseline (lag_1) |
|---:|---:|---:|
| 10 | 37.41% | 37.67% |
| 20 | 59.77% | 60.07% |
| 50 | 89.80% | 90.12% |

Hasil menunjukkan bahwa penambahan jumlah zona yang dipilih meningkatkan pickup coverage secara signifikan. Pemilihan 10 zona mampu mencakup sekitar 37% pickup aktual, sedangkan pemilihan 20 zona mencakup sekitar 60%. Ketika jumlah zona yang dipilih meningkat menjadi 50, coverage mencapai sekitar 90%.

Namun, strategi prioritas SVM belum menunjukkan performa yang lebih baik dibandingkan baseline lag-1. Baseline menghasilkan coverage sedikit lebih tinggi pada seluruh nilai K yang diuji. Perbedaannya relatif kecil, sehingga strategi SVM menghasilkan coverage yang sebanding, tetapi belum menunjukkan peningkatan dibandingkan baseline dalam eksperimen ini.

Temuan ini menunjukkan bahwa performa klasifikasi dan efektivitas operasional dalam menentukan prioritas zona merupakan dua aspek evaluasi yang berbeda. Model dengan performa klasifikasi yang baik belum tentu menghasilkan ranking zona yang lebih efektif untuk alokasi sumber daya.

## 5. Diskusi Keseluruhan

Hasil evaluasi menunjukkan bahwa setiap pendekatan memiliki keunggulan berdasarkan tujuan prediksi masing-masing.

Model regresi dapat digunakan untuk menghasilkan estimasi jumlah pickup secara numerik dengan kebutuhan komputasi yang relatif rendah. Model time-series forecasting dirancang untuk menangkap pola temporal, dengan SARIMA menghasilkan performa validation city-level terbaik di antara model forecasting yang diuji. Model klasifikasi menghasilkan kategori demand, dengan optimized RBF SVM memberikan performa classification terbaik di antara konfigurasi SVM yang dievaluasi.

Meskipun demikian, model-model tersebut tidak dapat dibandingkan secara langsung menggunakan satu metrik karena memiliki target dan tujuan prediksi yang berbeda. MAE dan RMSE mengukur error prediksi numerik, sedangkan Macro F1 mengukur performa klasifikasi pada kategori demand.

Analisis Coverage@K juga menunjukkan bahwa performa prediksi tidak selalu secara langsung menghasilkan efektivitas operasional yang lebih tinggi. Dalam validation experiment, strategi prioritas SVM menghasilkan coverage yang sebanding dengan baseline lag-1, tetapi belum berhasil melampaui baseline.

Dengan demikian, pemilihan model perlu mempertimbangkan akurasi prediksi, efisiensi komputasi, interpretabilitas, tujuan prediksi, dan manfaat operasional yang dihasilkan.

---

## 6. Kesimpulan

Berdasarkan hasil validation, optimized RBF SVM merupakan model klasifikasi terbaik yang dievaluasi oleh Member 4, dengan Macro F1 sebesar 0.7558 dan accuracy sebesar 75.57%. SARIMA menghasilkan performa city-level terbaik di antara model forecasting yang diuji, dengan WAPE sebesar 7.64% dan R² sebesar 0.9578. Pada model regresi, Lasso awal menghasilkan MAE terendah, sedangkan tuned Ridge memberikan performa prediksi numerik yang kompetitif dengan waktu training lebih rendah.

Cross-model comparison menunjukkan bahwa pendekatan regresi, forecasting, dan klasifikasi memiliki kekuatan berdasarkan tujuan prediksi masing-masing. Perbedaan target dan metrik evaluasi menyebabkan performa ketiganya tidak dapat dibandingkan secara langsung menggunakan satu angka.

Selain itu, hasil Coverage@K menunjukkan bahwa strategi prioritas berbasis SVM belum menghasilkan peningkatan coverage dibandingkan baseline lag-1 pada validation data.

Secara keseluruhan, model terbaik tidak hanya ditentukan oleh performa prediksi, tetapi juga oleh efisiensi komputasi, interpretabilitas, tujuan penggunaan, dan efektivitasnya dalam mendukung pengambilan keputusan.
