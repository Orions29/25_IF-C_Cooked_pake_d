# 5. Memeriksa apakah nama kolom dan tipe data konsisten antarbulan

```python
# Simpan kolom dan tipe data acuan dari DataFrame bulan Januari / Bulan 1
expected_columns = df[1].columns
expected_dtypes = df[1].dtypes

print("Expected Column beserta Tipe Datanya:")
for col, dtype in expected_dtypes.items():
    print(f"- {col}: {dtype}")
print("-" * 40)

expected_set = set(expected_columns)

# Cek tiap DataFrame di dictionary `df`
for i, data in df.items():
    actual_set = set(data.columns)
    actual_dtypes = data.dtypes
    
    # Cari kolom yang kurang atau kelebihan
    missing = expected_set - actual_set
    extra = actual_set - expected_set
    
    # Cek tipe data HANYA untuk kolom yang sama-sama ada (beririsan)
    common_cols = expected_set & actual_set
    dtype_mismatches = {}
    
    for col in common_cols:
        if expected_dtypes[col] != actual_dtypes[col]:
            dtype_mismatches[col] = (expected_dtypes[col], actual_dtypes[col])
    
    # Evaluasi hasil
    if not missing and not extra and not dtype_mismatches:
        print(f"Dataset Bulan {i}: Josjis, kolom dan tipe data sama persis! ")
    else:
        print(f"Dataset Bulan {i}: [Warning] - ada ketidaksesuaian!!!!!!!!!!!")
        if missing:
            print(f"   - Kurang (Missing): {missing}")
        if extra:
            print(f"   - Kelebihan (Extra): {extra}")
        if dtype_mismatches:
            print(f"   - Beda Tipe Data:")
            for col, (exp_type, act_type) in dtype_mismatches.items():
                print(f"       > Kolom '{col}' -> Seharusnya: {exp_type}, Ditemukan: {act_type}")
    # Jarak antar bulan
    print(">>> ") 
```

# 6. Missing Someone Special

 Memeriksa kolom yang seluruh nilainya kosong atau hanya tersedia pada bulan tertentu.

```python
# ============================================
# 6. Missing Value Check (Section 6)
# ============================================
print("=" * 60)
print("CEK MISSING VALUE PER BULAN")
print("=" * 60)

output_path = '../out/audit/missing_value.md'

missing_report = []
md_output = []

for i, data in df.items():
    print(f"\n{'='*20} DATASET BULAN {i} {'='*20}")
    
    # Jumlah missing per kolom
    missing = data.isnull().sum()
    missing_pct = (missing / len(data) * 100).round(2)

    # Display Sample Missingnya
    display(data.sample(5).isnull())
    
    # Ambil hanya kolom yang ada missing-nya
    missing_cols = missing[missing > 0]
    
    if len(missing_cols) > 0:
        report = pd.DataFrame({
            'bulan': i,
            'kolom': missing_cols.index,
            'jumlah_missing': missing_cols.values,
            'persentase_%': missing_pct[missing_cols.index].values
        })
        missing_report.append(report)
        display(report)

        
    else:
        print("✅ Tidak ada missing value pada semua kolom.")
    
    # Cek kolom yang SELURUHNYA kosong (100% missing)
    empty_cols = data.columns[data.isnull().all()].tolist()
    if empty_cols:
        print(f"⚠️  Kolom yang SELURUHNYA kosong di Bulan {i}: {empty_cols}")


# Gabungkan laporan semua bulan
if missing_report:
    missing_df = pd.concat(missing_report, ignore_index=True)
    print("\n" + "=" * 60)
    print("RINGKASAN SEMUA MISSING VALUE (SELURUH BULAN)")
    print("=" * 60)
    display(missing_df.sort_values(['bulan', 'jumlah_missing'], ascending=[True, False]))
    md_output.append(f'### RINGKASAN SEMUA MISSING VALUE (SELURUH BULAN)  {i}\n')
    md_output.append(missing_df.to_markdown(index=False))
    md_output.append('\n---\n')  # Garis pemisah antar bulan
else:
    print("\n✅ Semua dataset lengkap, tidak ada missing value.")


final_markdown = '\n'.join(md_output)

with open(output_path, 'w', encoding='utf-8') as f:
  f.write(final_markdown)
  
print('Berhasil disimpan ke hasil_dataset.md!')

# ============================================
# Cek kolom yang hanya ada di sebagian bulan (bukan semua)
# ============================================
print("\n" + "=" * 60)
print("CEK KOLOM YANG HANYA ADA DI PARAH BULAN (TIDAK KONSISTEN)")
print("=" * 60)

all_cols = set()
cols_per_bulan = {}

for i, data in df.items():
    cols_per_bulan[i] = set(data.columns)
    all_cols.update(data.columns)

for col in sorted(all_cols):
    bulan_ada = [i for i, cols in cols_per_bulan.items() if col in cols]
    if len(bulan_ada) < len(df):
        print(f"Kolom '{col}' hanya ada di bulan: {bulan_ada}")
```

# 7. Kenapa Dataset ini sesuai dengan kebutuhan latihan kedepannya?

Berdasarkan hasil audit di atas, dataset **NYC Yellow Taxi Trip Records Januari–Juni 2026** memenuhi syarat untuk digunakan sebagai data latihan (training) dan pemodelan di Fase 3–7, dengan alasan sebagai berikut:

1. **Cakupan waktu yang cukup untuk split kronologis**
   Tersedia 6 bulan data (Januari–Juni 2026), yang memungkinkan pembagian data secara berurutan waktu:
   - **Training:** Januari–April 2026
   - **Validation:** Mei 2026
   - **Testing:** Juni 2026
   Ini sesuai dengan aturan emas proyek: tidak ada random split atau k-fold acak.

2. **Volume data yang cukup besar**
   Total sekitar 22,8 juta baris dari enam file Parquet. Volume ini cukup untuk melatih berbagai model (Regresi, SVM, Time Series) serta untuk analisis spasial per zona per jam.

3. **Memiliki kolom waktu yang lengkap**
   Kolom `tpep_pickup_datetime` dan `tpep_dropoff_datetime` tersedia dengan tipe `datetime64[us]`, yang diperlukan untuk:
   - Pembuatan fitur temporal (`hour`, `day_of_week`, `month`, dll.)
   - Agregasi menjadi `demand_zone_hour` (pickup per zona per jam)
   - Analisis time series (Holt-Winters, SARIMA)

4. **Memiliki kolom lokasi untuk analisis spasial**
   `PULocationID` dan `DOLocationID` tersedia dan dapat diganti dengan koordinat centroid zona TLC (WGS84) untuk analisis hotspot (Moran's I, Getis-Ord Gi*) dan pembuatan fitur lokasi.

5. **Target dapat diturunkan secara jelas**
   Dari data ini dapat dibentuk target utama proyek: **`pickup_count` per zona per jam** (untuk regresi dan time series) serta **`demand_level`** (untuk klasifikasi SVM).

6. **Terdapat atribut pendukung analisis revenue (opsional)**
   Kolom tarif dan biaya (`fare_amount`, `total_amount`, `tip_amount`, `tolls_amount`, dll.) tersedia untuk analisis pendapatan di jalur Revenue, yang dipisah dari jalur Demand.

7. **Tersedia dokumentasi resmi dan data dictionary**
   Data dictionary dari NYC TLC tersedia di `datasets/datasets_docs/pdf/data_dictionary_trip_records_yellow.pdf`, sehingga definisi setiap kolom jelas dan terpercaya.

8. **Data 2016 tersedia sebagai pembanding opsional**
   Data Januari–Maret 2016 tersedia untuk analisis pembanding, namun harus diperlakukan terpisah dan tidak digabungkan ke dalam time series 2026 karena skema berbeda dan tidak membentuk deret waktu kontinu.
