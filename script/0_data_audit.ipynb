# 2. Audit Kelengkapan Dataset

Apakah dataset benar-benar lengkap 7 bulan januari sampai juni

```python
df = {}  # 1. Inisialisasi dictionary untuk menampung DataFrame tiap bulan

for i in range(1, 7):
    # Format {i:02d} otomatis membuat angka 1-9 jadi '01', '02', dst.
    file_path = rf"..\data\parquet\2026\yellow_tripdata_2026-{i:02d}.parquet"
    
    try:
        df[i] = pd.read_parquet(file_path)
        print(f"Dataset Bulan {i} berhasil dimuat.")
    except FileNotFoundError:
        print(f"Dataset Bulan {i} gaada!!")
```

# 5. Memeriksa apakah nama kolom dan tipe data konsisten antarbulan

```python
pick_columns = df[1].columns

print(f"Expected Column yang harusnya ada: \n{pick_columns}")

expected_set = set(pick_columns)

# 2. Cek tiap DataFrame di dictionary `df`
for i, data in df.items():
    actual_set = set(data.columns)
    
    # Cari kolom yang kurang atau kelebihan
    missing = expected_set - actual_set
    extra = actual_set - expected_set
    
    if not missing and not extra:
        print(f"Dataset Bulan {i}: Josjis, kolom lengkap ")
    else:
        print(f"Dataset Bulan {i}: [Warning] - ada ketidaksesuaian kolom!")
        if missing:
            print(f"   - Kurang (Missing): {missing}")
        if extra:
            print(f"   - Kelebihan (Extra): {extra}")
```
