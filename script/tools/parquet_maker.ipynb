# 5. Merge Each Tripdata with Zone Parquet

```python
# 1. Buat dictionary mapping dari coords_df yang udah kita buat tadi
lat_dict = df_zone_parquet.set_index('LocationID')['latitude'].to_dict()
lon_dict = df_zone_parquet.set_index('LocationID')['longitude'].to_dict()

# 2. Buat kolom koordinat penjemputan (Pick Up)
df['pickup_latitude'] = df['PULocationID'].map(lat_dict)
df['pickup_longitude'] = df['PULocationID'].map(lon_dict)

# 3. Buat kolom koordinat penurunan (Drop Off)
df['dropoff_latitude'] = df['DOLocationID'].map(lat_dict)
df['dropoff_longitude'] = df['DOLocationID'].map(lon_dict)

# 4. Hapus kolom ID lama (PULocationID & DOLocationID) kalau mau diganti total
df = df.drop(columns=['PULocationID', 'DOLocationID'])

# Cek hasil 5 baris pertama
print(df[['pickup_latitude', 'pickup_longitude', 'dropoff_latitude', 'dropoff_longitude']].head())
```
